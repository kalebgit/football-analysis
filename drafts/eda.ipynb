```python
#cuantas yardas permite el equipo defensivo por cuarto
import polars as pl
#penalty_team y penalty_yards
(pbp
    .filter(pl.col("defteam").is_not_null())
    .group_by("defteam", "qtr", "season").agg(
        pl.col("yards_gained").sum(),
        pl.when(pl.col("penalty_team") == pl.col("defteam"))
            .then(pl.col("penalty_yards"))
            .otherwise(0)
            .sum()
    ).sort(["defteam", "season", "qtr"], descending=[False, False, False]))
```
