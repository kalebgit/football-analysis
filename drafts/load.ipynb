```python
import polars as pl

BUCKET_NAME = "football-analysis-2026-kaleb"

def s3_key(*parts): 
    return "/".join(str(p).strip("/") for p in parts)

def load_to_s3(name, df):
    base_uri = f"s3://{BUCKET_NAME}"
    base = ["raw", name]
    if "season" in df.columns: 
        for season in df["season"].unique().sort():
            dataset = df.filter(pl.col("season") == season).drop("season")
            key = s3_key(*base, f"season={season}", "data.parquet")
            URI = "/".join([base_uri, key])
            print(URI)
    else: 
        key = s3_key(*base, "data.parquet")
        URI = "/".join([base_uri, key])
        print(URI)





datasets = {
    "pbp": pbp, 
    "schedules": schedules, 
    "rosters": rosters, 
    "player_stats": player_stats, 
    "contracts": contracts, 
    "participation": participation, 
    "ftn_charting": ftn_charting
}

for name, dataset in datasets.items(): 
    load_to_s3(name, dataset)
```
